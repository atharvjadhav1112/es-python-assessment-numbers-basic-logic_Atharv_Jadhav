## Question 1: Even or Odd

**Task:** Write a function that determines whether a number is even or odd.

**Approach:**
- Use the modulo operator (`%`) to check the remainder when the number is divided by 2.
- If `n % 2 == 0`, the number is even; otherwise it is odd.
- Edge cases handled:
  - Negative numbers (e.g., `-4` is even, `-7` is odd) — the modulo operator in Python works correctly here.
  - `0` is treated as even (mathematically correct).
  - Invalid input types (e.g., strings, `None`, floats that aren't whole numbers) raise a clear `TypeError`/`ValueError` instead of failing silently.


```python
def even_or_odd(n: int) -> str:
    """
    Determine whether a number is even or odd.

    Parameters
    ----------
    n : int
        The integer to check.

    Returns
    -------
    str
        "Even" if n is divisible by 2, otherwise "Odd".

    Raises
    ------
    TypeError
        If n is not an integer (or a whole-number float).
    """
    # Basic input validation / edge case handling
    if isinstance(n, bool):
        # bool is a subclass of int in Python; explicitly reject it
        raise TypeError("Boolean values are not valid numeric input.")
    if isinstance(n, float):
        if not n.is_integer():
            raise ValueError("Only whole numbers are supported (got a non-integer float).")
        n = int(n)
    if not isinstance(n, int):
        raise TypeError(f"Expected an int, got {type(n).__name__}.")

    return "Even" if n % 2 == 0 else "Odd"
```

### Test Cases — Question 1

```python
# Sample / edge case tests for even_or_odd
test_cases = [4, 7, 0, -4, -7, 100, 2.0]

for value in test_cases:
    print(f"even_or_odd({value!r}) -> {even_or_odd(value)}")


```

```python
# Invalid input handling demo
invalid_inputs = ["five", None, 3.5, True]

for value in invalid_inputs:
    try:
        result = even_or_odd(value)
        print(f"even_or_odd({value!r}) -> {result}")
    except (TypeError, ValueError) as e:
        print(f"even_or_odd({value!r}) raised {type(e).__name__}: {e}")
```

## Question 2: Sum of a List

**Task:** Write a function that calculates the total sum of a list of numbers.

**Approach:**
- Iterate through the list using a `for` loop and accumulate the total in a running `total` variable.
- (Deliberately avoiding the built-in `sum()` to demonstrate manual looping/accumulation logic, as requested by the assignment.)
- Edge cases handled:
  - Empty list -> returns `0`.
  - `None` or non-list input -> raises a clear `TypeError`.
  - List containing non-numeric elements -> raises a clear `TypeError` identifying the offending element.
  - Works with a mix of `int` and `float` values.


```python
def add_all(numbers: list) -> int:
    """
    Calculate the total sum of a list of numbers.

    Parameters
    ----------
    numbers : list
        A list of numeric values (int or float).

    Returns
    -------
    int (or float, if any element is a float)
        The sum of all elements in the list. Returns 0 for an empty list.

    Raises
    ------
    TypeError
        If `numbers` is not a list, or if it contains non-numeric elements.
    """
    if not isinstance(numbers, list):
        raise TypeError(f"Expected a list, got {type(numbers).__name__}.")

    total = 0
    for index, value in enumerate(numbers):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise TypeError(
                f"Element at index {index} is not a number: {value!r}"
            )
        total += value

    return total
```

### Test Cases — Question 2

```python
# Sample / edge case tests for add_all
print(add_all([1, 2, 3, 4, 5]))   
print(add_all([]))                
print(add_all([10, -5, 2.5]))    
print(add_all([100]))             
```

```python
# Invalid input handling demo
try:
    add_all("not a list")
except TypeError as e:
    print(f"Raised {type(e).__name__}: {e}")

try:
    add_all([1, 2, "three", 4])
except TypeError as e:
    print(f"Raised {type(e).__name__}: {e}")
```
